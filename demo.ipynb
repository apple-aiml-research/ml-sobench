```python
from openai import OpenAI
import json
import tqdm

model_name: str = "gpt-4o-mini"
OPENAI_API_KEY = ""
client = OpenAI(api_key=OPENAI_API_KEY)
```

```python
# Generating eval jsonl file first through `convert_to_eval_jsonl.py`.
label_path = "data/so_bench_eval.jsonl"
with open(label_path, "r", encoding="utf-8") as f:
    labels = [json.loads(line) for line in f]
print(f"Loading {len(labels)} data samples.")
```

```python
# Running inferences. You can either directly adding prediction 
# results directly back to labels for evaluation.
for sample in tqdm.tqdm(labels):
  response = client.chat.completions.create(
                  model=model_name,
                  messages=sample["messages"],
              )
  content = response.choices[0].message.content
  sample["generation"] = content
```

```python
# Compute evalaution metrics.
from eval import metric_fn
labels_with_predictions = labels
print(metric_fn(labels))
```
